**15.** Basado en los datos, calcula la probabilidad de que un país seleccionado al azar sea de 'Sudamérica' Y tenga una `Categoria_Desarrollo` de 'Medio-Alto'.

```python
#Esto se trata de un cálculo de probabilidad sencillo. Para ello, aplicamos la fórmula de:
#P(A) = n(A)/n(S)
#Donde n(A) es el número de casos favorables y n(S) es el número de casos posibles.
#En este caso, n(A) es número de países que cumplen la condición lógica:
#(Ser de Sudamérica) y (Categoría_Desarrollo == Medio Alto)
#n(S) es el número total de paises

#Ordenamos primero los datos en una tabla para apreciar esto de una mejor manera
print(df_desarrollo[['Pais','Categoria_Desarrollo','Continente']])
#Luego hallamos los datos útiles para el cálculo de probabilidad:
#Primero, aplicamos un filtro a la tabla para hallar los paises que cumplen la condición:
paises_condicion = df_desarrollo[(df_desarrollo['Continente'] == 'Sudamérica') & (df_desarrollo['Categoria_Desarrollo'] == 'Medio-Alto')]
#Ahora hallamos la cantiadad de paises que cumplen la condición
cant_paises_condicion = len(paises_condicion)
print("La cantidad de paises que cumplen la condicion es:", cant_paises_condicion)
#Luego, la cantidad total de paises
paises_totales= len(df_desarrollo['Pais'])
print("La cantidad total de paises es:", paises_totales)
#Ahora es solamente aplicar la fórmula de probabilidad
probabilidad_preg15 = (cant_paises_condicion / paises_totales) * 100
print("La probabilidad de que un país sea de Sudamérica y tenga una categoría de desarrollo Medio-Alto es:")
print(int(probabilidad_preg15), "%")
```
