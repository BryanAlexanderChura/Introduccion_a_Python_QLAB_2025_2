## Sección 1: NumPy - Manipulación Avanzada de Arrays

**1.1.** Crea un array de NumPy llamado `vector_base` que contenga una secuencia de números desde 10 hasta 115 (inclusive), con un paso de 3 entre cada número.

```python
matriz_C = matriz_A * matriz_B
print(matriz_C)
```

**7.** Crea una nueva columna condicional `Categoria_Desarrollo`. Clasifica a los países según las siguientes reglas:

**9.** Utilizando `groupby()`, calcula para cada `Continente`:
- La **media** de `Esperanza_de_vida`.
- El **máximo** de `Indice_Gini`.
- El **conteo** de países.

```python
# Media
df_desarrollo_media = df_desarrollo.groupby("Continente")["Esperanza_de_vida"].mean().reset_index()
df_desarrollo_media = df_desarrollo_media.round(2)
df_desarrollo_media
```

```python
# Máximo
df_desarrollo_max = df_desarrollo.groupby("Continente")["Indice_Gini"].max().reset_index(name = "Indice_Gini_Max")
df_desarrollo_max
```

```python
# Conteo
df_desarrollo_conteo = df_desarrollo.groupby("Continente")["Pais"].count().reset_index(name = "Número de Países")
df_desarrollo_conteo = df_desarrollo_conteo.sort_values(by="Número de Países", ascending=False)
df_desarrollo_conteo
```

## Sección 3: Visualización de Datos

**10.** Crea un gráfico de barras horizontales (`barh`) que muestre el `Indice_Gini` de cada país. Ordena los países de menor a mayor desigualdad (Gini) para una mejor visualización. Asegúrate de que el gráfico tenga un título y etiquetas adecuadas.

```python
import matplotlib.pyplot as plt
df_Gini = df_desarrollo.sort_values(by="Indice_Gini", ascending=True)
fig, ax = plt.subplots(figsize = (10,12))
ax.barh(df_Gini["Pais"], df_Gini["Indice_Gini"], color="darkblue" ) 
ax.set_xlabel("Indice Gini")
ax.set_title("Clasificación de países según su Índice Gini")
plt.show()
```
